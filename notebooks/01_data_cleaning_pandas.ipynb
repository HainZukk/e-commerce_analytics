# 3. Order Items


```python
df_orders = pd.read_csv("../data/raw/order_items.csv")
```

```python
df_orders.head()
```

```python
df_orders.info()
```

```python
df_orders = (
    df_orders.groupby(["order_id", "product_id"], as_index=False)
    .agg(
        {
            "quantity": "sum",
            "unit_price": "mean",
            "discount_percentage": "mean",
            "discount_amount": "sum",
            "gross_sales": "sum",
            "tax_amount": "sum",
            "shipping_cost": "sum",
            "net_sales": "sum",
            "product_cost": "sum",
            "profit": "sum",
        }
    )
)
```

```python
cols_moneda = [
    "discount_amount",
    "gross_sales",
    "tax_amount",
    "shipping_cost",
    "net_sales",
    "product_cost",
    "profit",
]

df_orders[cols_moneda] = df_orders[cols_moneda].round(2)
```

4. Optimizacion de datos (RAM BIG DATA)


```python
df_orders["order_id"] = df_orders["order_id"].astype(str)
df_orders["product_id"] = df_orders["product_id"].astype(str)
```

```python
df_orders.to_csv("../data/processed/order_items_clean.csv", index=False)
```

# 4. Ecommerce Sales Analytics


```python
df_sales = pd.read_csv("../data/raw/ecommerce_sales_customer_analytics.csv")
```

1. Conversion de Horas y Fechas


```python
df_sales["order_datetime"] = pd.to_datetime(df_sales["order_date"] + " " + df_sales["order_time"], errors="coerce")
df_sales["order_date"] = pd.to_datetime(df_sales["order_date"])
```

2. Tratamiento e imputación explícita de valores nulos


```python
df_sales["return_status"] = df_sales["return_status"].fillna("No Returned")
df_sales["return_reason"] = df_sales["return_reason"].fillna("N/A")
df_sales["campaign_name"] = df_sales["campaign_name"].fillna("Organic / None")
df_sales["coupon_code"] = df_sales["coupon_code"].fillna("No Coupon")
df_sales["customer_review"] = df_sales["customer_review"].fillna("No Review Provided")
df_sales["review_sentiment"] = df_sales["review_sentiment"].fillna("No Review")
```

3. Normalización de cadenas de texto y código postal


```python
df_sales["customer_postal_code"] = df_sales["customer_postal_code"].astype(str)
cols_str_sales = ["customer_name", "customer_city", "customer_state"]
for col in cols_str_sales:
    df_sales[col] = df_sales[col].astype(str).str.strip()
```

```python
# Definir tasas de cambio aproximadas o estándar a USD
exchange_rates = {
    'USD': 1.00,
    'EUR': 1.08, 
    'GBP': 1.27,  
    'CAD': 0.74,  
    'AUD': 0.65,  
    'INR': 0.012, # 1 INR = 0.012 USD
    'AED': 0.27   
}

# Crear la columna de tasa de cambio y convertir las ventas netas a USD
df_sales['exchange_rate'] = df_sales['currency'].map(exchange_rates)
df_sales['net_sales_usd'] = (df_sales['net_sales'] / df_sales['exchange_rate']).round(2)
df_sales['profit_usd'] = (df_sales['profit'] / df_sales['exchange_rate']).round(2)

print("¡Conversión a USD realizada con éxito!")
df_sales[['order_id', 'currency', 'net_sales', 'exchange_rate', 'net_sales_usd']].head()
```

```python
# 1. Asegurar limpieza y formato numérico estricto en net_sales y profit
cols_a_limpiar = ['net_sales', 'gross_sales', 'profit', 'tax_amount', 'shipping_cost']

for col in cols_a_limpiar:
    if col in df_sales.columns:
        # Convertir a numérico forzando errores a NaN y luego redondear a 2 decimales exactos
        df_sales[col] = pd.to_numeric(df_sales[col], errors='coerce').round(2)

# 2. Crear la columna en Dólares (USD) ya limpia y redondeada
# (Si ya tienes el diccionario de divisas aplicado)
if 'currency' in df_sales.columns and 'exchange_rate' in df_sales.columns:
    df_sales['net_sales_usd'] = (df_sales['net_sales'] / df_sales['exchange_rate']).round(2)
```

```python
df_sales.to_csv("../data/processed/ecommerce_sales_analytics_clean.csv", index=False)
```
